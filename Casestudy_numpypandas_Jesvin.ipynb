```python
# 1. Import pandas and numpy
import pandas as pd
import numpy as np


# 2. Load the dataset using read_csv()
df = pd.read_csv(
    "https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv"
)
# 3. Display basic information

# Display the number of rows and columns
print("number of rows and columns \n",df.shape)

# Display the first 5 rows
print(" the first 5 rows \n",df.head())

# Display all column names
print("all column names \n",df.columns)

# Display the data types of each column
print("the data types of each column \n",df.dtypes)

# Display the number of missing values in each column
print("the number of missing values in each column\n",df.isnull().sum())

```

```python
# 4. Check and remove duplicates

# Count the number of duplicate rows
print("Duplicates:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()
```

```python

# 5. Rename the Title column to Movie_Title

df.rename(columns={"Title": "Movie_Title"}, inplace=True)
```

```python
# 6. Get unique genres

# Display all unique values in the Genre column
print("all unique values in the Genre column\n",df["Genre"].unique())
```

```python
# 7. Find the number of genres for each movie
df["Genre_Count"] = df["Genre"].str.split(",").str.len()
df[["Genre","Genre_Count"]]
```

```python

# 8. Handle missing values

# Replace missing Revenue values with the median revenue
df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(
    df["Revenue (Millions)"].median()
)

# Replace missing Metascore values with the median metascore
df["Metascore"] = df["Metascore"].fillna(
    df["Metascore"].median()
)

```

```python
# 9. Find the top 5 movies by Rating

# Sort movies by Rating from highest to lowest
# Then select the first 5 movies

print(df.sort_values("Rating", ascending=False).head(5)[["Movie_Title", "Rating"]])
```

```python
# 10. Movie with highest revenue

highest = df.sort_values('Revenue (Millions)', ascending=False).head(1)
print("Movie with highest and lowest revenue: ")

print(highest[['Movie_Title', 'Revenue (Millions)']])
print()

# Movie with lowest revenue.

lowest = df.sort_values('Revenue (Millions)').head(1)
print("Movie with highest and lowest revenue: ")

print(lowest[['Movie_Title', 'Revenue (Millions)']])
print()
```

```python
# 11. Find the average rating for each director
print(df.groupby("Director")["Rating"].mean())
```

```python
# 12. Count the number of movies in each year
print( df.groupby("Year").size())
```

```python
# 13. Select movies with rating greater than 8
# and revenue greater than 100 million
print(df[
    (df["Rating"] > 8) &
    (df["Revenue (Millions)"] > 100)
])
```

```python
# 14. Find average Rating and Revenue for each director
print(df.groupby("Director")[["Rating", "Revenue (Millions)"]].mean())
```

```python
# 15. Count movies and find average rating for each year
year_group = df.groupby('Year')[['Rating']].agg(['count', 'mean']) # Performs multiple calculation on 'Rating' column.
print("Group by Year: count, avg Rating: ")
print()
print(year_group)
print()
```

```python
# 16.Calculate Profitability
df["Profitability"] = df["Revenue (Millions)"] / df["Runtime (Minutes)"]

# Display top 10 most profitable movies
top10 = df.sort_values("Profitability", ascending=False).head(10)
print(top10[["Movie_Title", "Profitability"]])
```

```python
# 17. Create rating categories
def rating_category(rating):
    if rating < 6:
        return "Poor"
    elif rating < 7.5:
        return "Average"
    else:
        return "Excellent"

# Apply the function
df["Rating_Category"] = df["Rating"].apply(rating_category)

# Display the result
print(df[["Movie_Title", "Rating", "Rating_Category"]])
```

```python
# 10. Create a function to calculate movie score
def movie_score(row):
    return row["Rating"] * 10 + row["Revenue (Millions)"]

# Apply the function row by row
df["Score"] = df.apply(movie_score, axis=1)

# Display movie name and score
print(df[["Movie_Title", "Score"]])
```
